# 11 · The entry-level reality check

Do "junior" jobs really let newcomers in? From 2025 on (when levels and experience are usable): how many junior and intern postings ask for several years of experience, how many require a degree, and do they ask for as many technologies as senior ones?

**Method.** To avoid circular logic, "entry" here counts only postings whose level came from the **title or the board's label**, not from the experience figure (that fallback labels anything up to a year as junior by construction). The degree is the *lowest* the posting accepts, read from the board's education tags (filled for part of the feed, see section 3). The technology comparison is inside LinkedIn, whose descriptions keep the same length.

```python
from vnjobs.analysis import core as a
from vnjobs.analysis import entry_level as e

a.apply_style()
con = a.connect()
df = a.load_features(con)
```

## 1. Junior postings that ask for experience

Entry-level postings (title or label) that state a required experience, grouped by the years they ask for.

```python
entry = e.entry_with_experience(df)
buckets = e.experience_buckets(entry)
a.save_table(buckets, "11-entry-required-experience")
print(f"{len(entry):,} entry-level postings state an experience requirement")
buckets
```

```python
by_role = e.entry_by_role(entry)
a.save_table(by_role, "11-entry-experience-by-role")
by_role
```

## 2. Degree requirements by level

Minimum degree among postings that state one (2025 on), by seniority level.

```python
degrees = e.degree_by_level(df)
a.save_table(degrees, "11-degree-by-level")
degrees
```

## 3. Who states a degree

Share of postings that state a degree, per year and source (2024 on). Read the degree shares above with this coverage in mind.

```python
coverage = e.degree_coverage(df)
a.save_table(coverage, "11-degree-coverage")
coverage
```

## 4. Do junior postings ask for as much?

Average technologies named per posting by level, inside LinkedIn, 2025 on (postings with a description).

```python
techs = e.technologies_by_level(df)
a.save_table(techs, "11-technologies-by-level")
techs
```

## Findings

1. **About one in three "junior" postings asks for three or more years.** Of 4,130 entry-level postings (level from the title or the board's label) that state an experience requirement, 32.7 % ask for up to a year, 35.3 % for two years, 16.1 % for three and 16.0 % for four or more. So 67 % ask for two years or more and 32 % for three or more. The experience comes from the boards' structured fields, which move in steps (12, 24, 36, 60 months), and I have not read individual postings.
2. **Backend and DevOps are the worst offenders.** Share of entry-level postings asking for three years or more: backend 45.1 %, DevOps 43.5 %, full-stack 40.2 %, AI/ML 36.1 %, data and QA 29.7 %, embedded 25.0 %, security 24.6 %, general software 24.0 %, IT support 19.0 %.
3. **A degree is not a barrier specific to juniors.** Among postings that state a degree (2025 on), 73.7 % of junior postings require a bachelor's, against 81.5 % of senior and 60.8 % of mid-level ones; 23 % of junior and intern postings accept an associate degree or certificate. Coverage is uneven: CareerBuilder and CareerViet state a degree on 98–100 % of postings in 2025–26, LinkedIn on 78–84 %, and the other boards on none.
4. **Junior postings ask for fewer technologies, but still many.** Inside LinkedIn, junior postings name 6.1 technologies on average against 8.7 for senior; 54 % of junior postings name five or more (interns 40 %, seniors 75 %).

**What this notebook cannot say:** whether employers actually reject candidates with less experience than they list, or whether "junior" with "3 years" is an honest requirement or a copied template.
