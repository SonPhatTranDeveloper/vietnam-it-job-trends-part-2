# 14 · Benefits and perks (CareerBuilder only)

What do IT employers say they offer: insurance, training, bonus, a laptop?

**Method.** Benefit tags exist for about 90 % of 2021–22 postings but only 15–20 % later, and which boards fill the tag changes by year, so a cross-source trend would mostly show who fills the tag. This notebook stays **inside CareerBuilder**, shows the coverage first, and takes every share over the CareerBuilder postings that carry benefit tags. Tag names in Vietnamese and English are normalised to eleven benefits (`taxonomy.benefits_of`).

```python
from vnjobs.analysis import benefits as b
from vnjobs.analysis import core as a

a.apply_style()
con = a.connect()
df = a.load_features(con)
```

## 1. Coverage

Share of CareerBuilder postings that carry benefit tags, per year.

```python
coverage = b.coverage_by_year(df)
a.save_table(coverage, "14-benefit-coverage")
coverage
```

## 2. What is offered

Share of tagged CareerBuilder postings offering each benefit, per year.

```python
table = b.by_year_table(df)
a.save_table(table, "14-benefits-by-year")
table
```

```python
series = b.benefit_series(df)
fig = a.trend_lines(
    series, "Benefits offered, CareerBuilder postings with benefit tags", "% of tagged postings"
)
a.save_chart(fig, "14-benefits-trend")
fig
```

## Findings

1. **Within CareerBuilder the benefit tags are almost always filled** (91–98 % of postings in every year, because employers tick them from a fixed menu), so a CareerBuilder-only trend is valid. The low coverage seen across all sources is other boards not filling the tag.
2. **Some benefits are near-universal.** Insurance 97–99 % in every year; training 89 % (2020) to 96 %; salary review 91–96 %; bonus 88 % (2020) to 96 % (2026).
3. **Several have become more common.** Health care 73 % (2020) to 92 % (2026); travel 75 % to 89 %; annual leave 60 % (2021–22) to 78 % (2025–26); allowance 47 % (2021) to 65 %; **laptop 33 % (2021) to 51 % (2026)**; uniform 28 % to 50 %; sport club 14 % to 25 %.
4. **Caveat.** These are tags on a menu, ticked by the employer: they show what employers say they offer, not what it is worth, and a change in the board's menu could move a number. IT postings only.

**What this notebook cannot say:** what employers outside CareerBuilder offer, or anything about perks that are not on the menu (remote days, stock, a 13th-month salary).
